1. Write a function to print Unusual words from Gutenberg corpus. Print a list of all
stop words



```python
import nltk
nltk.download('words')

def unusual_words(text):
  text_vocab = set(w.lower() for w in text if w.isalpha())
  english_vocab = set(w.lower() for w in nltk.corpus.words.words())
  unusual = text_vocab.difference(english_vocab)
  return sorted(unusual)

nltk.download('gutenberg')
unusual_words(nltk.corpus.gutenberg.words('austen-emma.txt'))
```

3. Use beautifulsoup package to read a web page and create a frequency distribution of
most frequent words

```python
from urllib import request
from bs4 import BeautifulSoup

def freq_words(url, freqdist, n):
    html = request.urlopen(url).read().decode('utf8')
    raw = BeautifulSoup(html, 'html.parser').get_text()
    for word in word_tokenize(raw):
        freqdist[word.lower()] += 1
    result = []
    for word, count in freqdist.most_common(n):
        result = result + [word]
    print(result)
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk import word_tokenize

# Using an active alternative URL for the US Constitution
constitution = "https://www.gutenberg.org/cache/epub/5/pg5.txt"

fd = nltk.FreqDist()
freq_words(constitution, fd, 30)
```

4. Write code to develop a Look Up Tagger

```python
import nltk
nltk.download('punkt')
nltk.download('brown')
from nltk.corpus import brown

# Get the words and tagged sentences from the brown corpus
brown_words = brown.words(categories='news')
brown_tagged_sents = brown.tagged_sents(categories='news')

# Create a frequency distribution of words
fd = nltk.FreqDist(brown_words)
# Create a conditional frequency distribution of words and their tags
cfd = nltk.ConditionalFreqDist(brown.tagged_words(categories='news'))

# Get the 100 most frequent words
most_freq_words = fd.most_common(100)

# Find the most likely tag for each of these 100 words
likely_tags = dict((word, cfd[word].max()) for (word, _) in most_freq_words)

# Define a baseline backoff tagger for words not in our lookup table
backoff_tagger = nltk.DefaultTagger('NN')

# Define the lookup tagger using UnigramTagger
baseline_tagger = nltk.UnigramTagger(model=likely_tags, backoff=backoff_tagger)

# Evaluate the tagger
accuracy = baseline_tagger.accuracy(brown_tagged_sents)
print(f"Tagger Accuracy: {accuracy:.4f}")
```

8. Use Wordnet to extract various lexical relations

```python
import nltk
from nltk.corpus import wordnet

# Download the necessary WordNet data
nltk.download('wordnet')

def get_lexical_relations(word):
    synonyms = []
    antonyms = []
    hyponyms = []
    hypernyms = []

    for synset in wordnet.synsets(word):
        synonyms.extend(synset.lemma_names())
        antonyms.extend(lemma.antonyms()[0].name() for lemma in synset.lemmas() if lemma.antonyms())
        hyponyms.extend(hypo.name() for hyper in synset.hypernyms() for hypo in hyper.hyponyms())
        hypernyms.extend(hyper.name() for hyper in synset.hyponyms() for hyper in hyper.hypernyms())

    return list(set(synonyms)), list(set(antonyms)), list(set(hyponyms)), list(set(hypernyms))

word_to_lookup = "car"
synonyms, antonyms, hyponyms, hypernyms = get_lexical_relations(word_to_lookup)
print(f"Word: {word_to_lookup}")
print(f"Synonyms: {', '.join(synonyms)}")
print(f"Antonyms: {', '.join(antonyms)}")
print(f"Hyponyms: {', '.join(hyponyms)}")
print(f"Hypernyms: {', '.join(hypernyms)}")
```

10. Write a function to perform Unigram Chunking

```python
import nltk

class UnigramChunker(nltk.ChunkParserI):
    def __init__(self, train_sents):
        train_data = [[(t,c) for w,t,c in nltk.chunk.tree2conlltags(sent)]
                      for sent in train_sents]
        self.tagger = nltk.UnigramTagger(train_data)

    def parse(self, sentence):
        pos_tags = [pos for (word,pos) in sentence]
        tagged_pos_tags = self.tagger.tag(pos_tags)
        chunktags = [chunktag for (pos, chunktag) in tagged_pos_tags]
        conlltags = [(word, pos, chunktag) for ((word,pos),chunktag)
                     in zip(sentence, chunktags)]
        return nltk.chunk.conlltags2tree(conlltags)
```

```python
import nltk
from nltk.corpus import conll2000
from nltk.chunk import conlltags2tree, tree2conlltags
from nltk.chunk import ne_chunk
from nltk import pos_tag
```

```python
import nltk
nltk.download('conll2000')

test_sents = conll2000.chunked_sents('test.txt', chunk_types=['NP'])
train_sents = conll2000.chunked_sents('train.txt', chunk_types=['NP'])
unigram_chunker = UnigramChunker(train_sents)
print(unigram_chunker.evaluate(test_sents))
```

```python
postags = sorted(set(pos for sent in train_sents
                     for (word,pos) in sent.leaves()))
print(unigram_chunker.tagger.tag(postags))
```

15. Use a simple Context Free Grammar to develop a recursive descent grammar

```python
grammar1 = nltk.CFG.fromstring("""
  S -> NP VP
  VP -> V NP | V NP PP
  PP -> P NP
  V -> "saw" | "ate" | "walked"
  NP -> "John" | "Mary" | "Bob" | Det N | Det N PP
  Det -> "a" | "an" | "the" | "my"
  N -> "man" | "dog" | "cat" | "telescope" | "park" | "biscuit" | "while"
  P -> "in" | "on" | "by" | "with" | "it"
  """)
```

```python
sent = "Mary saw Bob".split()
rd_parser = nltk.RecursiveDescentParser(grammar1)
for tree in rd_parser.parse(sent):
  print(tree)
```

```python
sent2 = "Mary saw the dog in the park ".split()
rd_parser = nltk.RecursiveDescentParser(grammar1)
for tree in rd_parser.parse(sent2):
   print(tree)
```
